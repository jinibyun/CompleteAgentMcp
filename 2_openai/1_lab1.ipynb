```python
# The imports

from dotenv import load_dotenv
from agents import Agent, Runner, trace

```

```python
# The usual starting point

load_dotenv(override=True)
```

```python

# Make an agent with name, instructions, model  (모델은 어느 모델을 사용해도 가능하다)

agent = Agent(name="Jokester", instructions="You are a joke teller", model="gpt-4o-mini")
```

```python
# Run the joke with Runner.run(agent, prompt) then print final_output
# 역으로 알 수 있는 것: Runner.run() 은 비동기 함수라는 것, 그리고 결과 객체에는 final_output 속성이 있다는 것
# coroutine: 비동기 함수는 일반적으로 coroutine이라고 불리는 특별한 객체를 반환합니다. 이 객체는 나중에 실행될 수 있는 코드 블록을 나타냅니다. Runner.run()이 비동기 함수이므로, 이를 호출하면 coroutine 객체가 반환됩니다. 이 객체는 실제로 실행되기 전까지는 결과를 제공하지 않습니다.
with trace("Telling a joke"):
    result = await Runner.run(agent, "Tell a joke about Autonomous AI Agents")
    print(result.final_output)
```

#### 위의 with trace 를 통해 openai 와의 interaction 을 log 로 확인 할 수 있다
